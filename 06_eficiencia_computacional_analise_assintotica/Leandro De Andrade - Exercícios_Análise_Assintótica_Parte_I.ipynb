#Exercício 02

Insira o código anterior dentro de uma função e utilize time para medir o tempo de execução da função exemplo

    def funcaoExemplo():
      # Simula algum processamento
      for _ in range(1000000):
        pass

##Solução

````python
import math

def funcao(n):
    
    inicio = time.time()
    s = 0
    x = math.factorial(n)
    for i in range(1 , x):
        s += i
    fim = time.time()
    return s, f"Tempo decorrido:{fim-inicio:.6f}s."

print(funcao(10))
````

#Breve explicação sobre números aleatórios em python

A **`random()`** é uma função do módulo **`random`** do Python que devolve um **float pseudoaleatório no intervalo [0.0, 1.0)**.
Uso básico:

```python
import random
x = random.random()   # ex.: 0.7315...
```

Pontos rápidos:

* Os números são **pseudoaleatórios** (gerados por algoritmo); para reprodutibilidade use **`random.seed(42)`**.
* Variantes úteis no mesmo módulo:
  
  * `randint(a, b)`: inteiro entre *a* e *b* (inclusive).
  * `randrange(início, fim, passo)`: inteiro em um range.
  * `uniform(a, b)`: float entre *a* e *b*.
  * `choice(seq)`: escolhe 1 item de uma sequência.
  * `shuffle(lista)`: embaralha uma lista in place.
  * `sample(seq, k)`: amostra *k* itens sem reposição.


#Exercício 03


Crie uma função que receba uma lista vazia de tamanho n=10 e preencha com valores aleatórios entre 1000000 e 10000000.

Execute a função 10 vezes, medindo o tempo de execução  em cada execução.

Observe se há uma alteração relevante no tempo!

Utilize random..randint()

    random.randint(1000000, 10000000)

##Solução

````python
import random

lista = [None]*10

def func_03(lista):
    inicio = time.time()
    for i in range(len(lista)):
        lista[i] = random.randint(1000000, 10000000)
    fim = time.time()
    t_exec = fim - inicio
    return lista, t_exec


execucoes_03 = []
tamanhos_n_03 = [] 

for l in range(10):
    funcao, tempo = func_03(lista)
    execucoes_03.append(tempo)
    tamanhos_n_03.append(f"Exec {l+1}")
    display(funcao, tempo)
````

#Exercício 04


Crie uma função que receba uma lista vazia de tamanho n e preencha com valores aleatórios entre 1 e 1000.

Execute a função 5 vezes, medindo o tempo de execução  em cada execução.

IMPORTANTE: O valor inicial de n será 10 e, a cada execução, o valor de n será atualizado como n *=10.

Observe se há uma alteração relevante no tempo!

#Solução

````python
def func_04(n):
    inicio = time.time()
    lista = [None] * n
    for i in range(len(lista)):
        lista[i] = random.randint(1,1000)
    fim = time.time()
    t_exec = fim-inicio
    return t_exec

n = 10

execucoes_04 = []
tamanhos_n_04 = []

for i in range(5):
    tempo = func_04(n)
    execucoes_04.append(tempo) 
    tamanhos_n_04.append(str(n))
    display(f"{tempo:.6f}s.")
    n *= 10
````

#Exercício 06

Implemente de forma iterativa o cálculo da série matemática abaixo.

    s = 1 + 2 + 3 + 4 +...+n

Teste para n = 1000, 10000, 100000, 1000000,10000000

Utilizando time(), retornar o tempo de execuçao para cada valoro de n.

Avalie o crescimento no tempo de execução

#Solução

````python
n = [1000, 10000, 100000, 1000000,10000000]
dic_iter={}
for i in n:
    inicio = time.time()
    s = 0
    for j in range(1, i+1):
        s += j
    fim = time.time()
    t_exec = fim - inicio
    dic_iter[i] =t_exec
    print(i, "->", t_exec)
    
````

A variação no tempo de execução ocorre em proporção direta ao aumento do valor de N. Como o algoritmo utiliza um laço iterativo para realizar a soma passo a passo, o esforço computacional cresce na mesma medida que o tamanho N. Ao multiplicarmos N por 10 a cada teste, observamos que o tempo de processamento também aumenta em aproximadamente 10 vezes. Esse comportamento ilustra a complexidade de tempo linear O(n).

#Exercício 07

Repita a atividade anterior mas usando a fórmula fechada da série

    s=1+2+3+4+...+n = n(n+1)/2

Avalie o crescimento no tempo de execução

#Solução

````python
n = [1000, 10000, 100000, 1000000,10000000]
dic_form={}
for i in n:
    inicio = time.time()
    s = 0
    s = i*(i+1)//2
    fim = time.time()
    t_exec = fim - inicio
    dic_form[i] = t_exec
    print(i,"->",t_exec)
````

Não há um aumento de tempo perceptível, independentemente do aumento de N. Como o algoritmo resolve o problema utilizando uma fórmula matemática, o processador executa a mesma quantidade de operações para qualquer valor da lista. O esforço computacional permanece igual e não acompanha o crescimento de N, ilustrando o comportamento da complexidade de tempo constante O(1).

Considere o cálculo do n-ésimo termo de uma Progressão Geométrica \(PG\). Podemos criar um algoritmo iterativo ou simplesmente usar a fórmula.

Por exemplo, para calcular o termo n=5 de uma PG com termo inicial 1 e razão 2, temos:

No caso iterativo:

a0=1

a1=2*1=2

a2=a1*2=2*2=4

a3=a2*2=4*2=8

a4=a3*2=16

a5=a4*2=32

Usando a fórmula da PG an=a0*(r^n)

a5=a0*(r^n)=1,(2^5)=32

Repita a comparação do exercício 5 para a versão iterativa e para a versão usando fórmula da PG. Considere a0=1 r=2 e n como sendo o valor de i variando no range() abaixo:

    for i in range(0,10000,10)



#Solução

````python
a1 = 1
r = 2
n = 10

def iter_pg(a1,r,n):
    inicio = time.time()
    pg=[]
    for _ in range(n):
        pg.append(a1)
        a1 *= r
    fim = time.time()
    t_exec = fim - inicio
    return  t_exec

tempo_pg_iter = []
for n in range(0,10000,10):
    iterativa = iter_pg(1,2,n)
    tempo_pg_iter.append(iterativa)
````

````python
a1 = 1
r = 2
n = 10

def mat_pg(a1,r,n):
    inicio = time.time()
    an=a1*(r**n)
    fim = time.time()
    t_exec = fim - inicio
    return  t_exec

tempo_pg_mat = []
for n in range(0,10000,10):
    formula = mat_pg(1,2,n)
    tempo_pg_mat.append(formula)
    
````

````python
import matplotlib.pyplot as plt

eixo_x = list(range(0, 10000, 10))

fig, ax = plt.subplots()

ax.scatter(eixo_x, tempo_pg_iter, marker=".",label="PG Iterativa")
ax.scatter(eixo_x, tempo_pg_mat, marker=".",label="PG Fórmula")

plt.title("Comparação de Desempenho: Iteração vs Fórmula")
plt.xlabel("Valor de N")
plt.ylabel("Tempo de Execução (segundos)")
ax.legend()

plt.show()
````

O gráfico demonstra a diferença entre as duas abordagens. O método iterativo mostra a tendência de crescimento linear O(n). Esse exercício mostra que a complexidade da modelagem matemática direta é computacionalmente superior à estrutura de repetição.
