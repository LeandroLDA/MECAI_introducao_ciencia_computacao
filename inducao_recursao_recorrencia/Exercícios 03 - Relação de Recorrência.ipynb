#Exercício 01
Implementar a relaçao de recorrência abaixo:

T(1) = 1

T(n) = 2T(n-1) para n>1

Medir o tempo de execução para entradas n com

    range(10, 1000, 10)

Plotar o tempo de execução para as entradas acima.

Dica para evitar problemas com as chamadas recursivas

    import sys

    sys.setrecursionlimit(10**6)

##Solução

````python
import sys
import time
import matplotlib.pyplot as plt

sys.setrecursionlimit(10**6)

def T(n):
    if n == 1:
        return 1
    return 2 * T(n - 1)

valores_n = list(range(10, 1000, 10))
tempos_execucao = []

inicio = time.time()

for n in valores_n:
    T(n)
    fim = time.time()
    tempo_decorrido = fim - inicio
    tempos_execucao.append(tempo_decorrido)

plt.figure(figsize=(10, 6))
plt.plot(valores_n, tempos_execucao, marker='o', linestyle='-', color='b')
plt.title('Tempo de Execução da Relação de Recorrência')
plt.xlabel('Tamanho da Entrada')
plt.ylabel('Tempo de Execução em segundos')
plt.grid(True)
plt.tight_layout()
plt.show()
````

<font color = 'orange'>O gráfico exibe um aumento de tempo linear de complexidade O(n), o que reflete a sobrecarga progressiva gerada pelo decaimento unitário T(n-1) das chamadas recursivas na memória.

#Exercício 02

Repetir a analise anterior para a função recursiva:

T(1) = 1
T(n) = T(n//2) +1

onde n//2 retorna a divisão inteira.

##Solução

````python
import sys
import time
import matplotlib.pyplot as plt

sys.setrecursionlimit(10**6)

def T(n):
    if n <= 1:
        return 1
    return T(n // 2) + 1

valores_n = list(range(10, 1000, 10))
tempos_execucao = []

inicio = time.time()

for n in valores_n:
    T(n)
    fim = time.time()
    tempo_decorrido = fim - inicio
    tempos_execucao.append(tempo_decorrido)

# 3. Plotar o tempo de execução
plt.figure(figsize=(10, 6))
plt.plot(valores_n, tempos_execucao, marker='o', linestyle='-', color='g')
plt.title('Tempo de Execução da Relação de Recorrência T(n) = T(n//2) + 1')
plt.xlabel('Tamanho da Entrada (n)')
plt.ylabel('Tempo de Execução (segundos)')
plt.grid(True)
plt.tight_layout()
plt.show()
````

<font color = "orange">O gráfico ilustra a complexidade logarítmica calculada pelo Método Mestre. Como o problema é sucessivamente dividido pela metade, a profundidade recursiva cai drasticamente mantendo o tempo de execução constante e estável.

#Exercício 03
Definda uma relação de recorrência para a série:

s=1+2+3+4...+n

Implemente uma função recursiva e repita a análise realizada nos exercícios anteriores.

##Solução

````python
import sys
import time
import matplotlib.pyplot as plt

sys.setrecursionlimit(10**6)

def soma_serie(n):
    if n <= 1:
        return 1
    return soma_serie(n - 1) + n

valores_n = list(range(10, 1000, 10))
tempos_execucao = []

inicio = time.time()
for n in valores_n:
    soma_serie(n)
    fim = time.time()
    tempo_decorrido = fim - inicio
    tempos_execucao.append(tempo_decorrido)

plt.figure(figsize=(10, 6))
plt.plot(valores_n, tempos_execucao, marker='o', linestyle='-', color='r')
plt.title('Tempo de Execução da Relação de Recorrência')
plt.xlabel('Tamanho da Entrada')
plt.ylabel('Tempo de Execução em segundos')
plt.grid(True)
plt.tight_layout()
plt.show()
````

<font color = "orange">O crescimento linear da curva confirma a complexidade do decaimento unitário das chamadas. Esse crescimento mostra a sobrecarga do sistema operacional que precisa alocar memória extra para conseguir empilhar as chamadas recursivas.

# Exercício 04 — Exponenciação: Ingênuo vs. Rápido
**Enunciado.** Compare dois algoritmos para \(a^n\): (1) ingênuo (\(n\) multiplicações), e (2) exponenciação por quadrados (divide & conquer).
  

*   (a)Defina a relação de recorrência.
*   (b)Avalie a complexidade usando o Teorema Mestre
*   (c)Implemente e exiba comparações de desempenho, por exemplo, fixando a e variando n.

Aqui está o **pseudocódigo** da função `pow_fast` (exponenciação por quadrados):

```
Procedimento PotenciaRapida(a, n)
  Entrada: número a, inteiro não negativo n
  Saída: a^n

  se n = 0 então
      retorne 1
  se n = 1 então
      retorne a

  half ← PotenciaRapida(a, ⌊n/2⌋)

  se n é par então
      retorne half × half
  senão
      retorne half × half × a
```


#Solução


````python
import sys
import time
import matplotlib.pyplot as plt

sys.setrecursionlimit(10**6)

def pow_naive(a, n):
    if n == 0:
        return 1
    if n == 1:
        return a
    return a * pow_naive(a, n - 1)

def pow_fast(a, n):
    if n == 0:
        return 1
    if n == 1:
        return a
    
    half = pow_fast(a, n // 2)
    
    if n % 2 == 0:
        return half * half
    else:
        return half * half * a

valores_n = list(range(10, 2000, 50))
tempos_naive = []
tempos_fast = []

a = 2

inicio_n = time.time()

for n in valores_n:
    pow_naive(a, n)
    fim_n = time.time()
    tempos_naive.append(fim_n - inicio_n)
 
inicio = time.time()   
for n in valores_n:
    pow_fast(a, n)
    fim = time.time()
    tempos_fast.append(fim - inicio)

plt.figure(figsize=(10, 6))
plt.plot(valores_n, tempos_naive, marker='o', linestyle='-', color='r', label='Ingênuo')
plt.plot(valores_n, tempos_fast, marker='s', linestyle='-', color='b', label='Rápido')
plt.title('Comparação de Desempenho')
plt.xlabel('Tamanho da Entrada')
plt.ylabel('Tempo de Execução em segundos')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
````

<font color = "orange">A curva vermelha exibe o custo linear e a sobrecarga de memória causados pelo intenso empilhamento de chamadas recursivas. Por outro lado, a linha azul do método rápido permanece constante pois o algoritmo fraciona o problema pela metade a cada passo, resultando em uma alta eficiência que é comprovada pelo Teorema Mestre.
