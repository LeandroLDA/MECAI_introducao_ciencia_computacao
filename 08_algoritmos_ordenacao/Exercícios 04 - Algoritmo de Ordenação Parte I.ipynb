Implementar o algoritmo bubble sort abaixo

Testar para ordenar listas geradas conforme abaixo:

    min,max = 10,100

    for i in range(5,20,5)
      L.append([random.randint(min, max) for _ in range(tamanho)])


### Solução

```python
import random
```

```python
def bubble_sort_otimizado(L):
    n = len(L)
    for i in range(n-1):
        troca = False
        for j in range(i+1, n):
            if L[i] > L[j]:
                L[j],L[i] = L[i],L[j]
                troca = True
        if ( troca==False):
            break          
    return L
```

```python
min, max = 10,100
L_bs=[]
for i in range(5,20,5):
   L_bs.append([random.randint(min, max) for _ in range(i)])
```

```python
for i in L_bs:
    print(f"Lista bruta:{i}")
    lista_buble = bubble_sort_otimizado(i)
    print(f"Lista ordenada:{lista_buble}\n")
```

## Exercício Prático 2

Repetir os testes do Exercício Prático 1 para a implementação do algoritmo insertion sort abaixo:

      INSERTION-SORT (L, n)
      Para j = 1 até n faça
          key = L[ j]
          i = j –1
          Enquanto (i > 0) E  L[i] >key faça
              L[i+1] =  L[i]
              i = i –1
          L[i+1] = key
      retorne L

### Solução

```python
def insertion_sort(L):
    n=len(L)
    for j in range(n):
        chave = L[j]
        i = j - 1
        while (i >= 0) and (L[i] > chave):
            L[i+1] = L[i]
            i = i - 1
        L[i+1] = chave
    return L
```

```python
min, max = 10,100
L_is=[]
for i in range(5,20,5):
    L_is.append([random.randint(min, max) for _ in range(i)])
```

```python
for i in L_is:
    print(f"Lista bruta:{i}")
    lista_insertion = insertion_sort(i)
    print(f"Lista ordenada:{lista_insertion}\n")
```

## Exercício Prático 3

Repetir os testes do Exercício Prático 1 para a implementação do algoritmo selection sort abaixo:

    algSelectionSort(L,n)
      Para i=0 até n-1 faça
        min = i
        Para j = i+1 até n faça
          Se(L[min]>L[j]) então
            min = j
        temp = L[i]
        L[i] = L[min]
        L[min] = temp

### Solução

```python
def selection_sort(L):
    n = len(L)
    for i in range(n-1):
        min = i
        for j in range(i+1,n):
            if(L[min]>L[j]):
                min = j
        temp = L[i]
        L[i] = L[min]
        L[min] = temp
    return L
```

```python
min, max = 10,100
L_ss=[]
for i in range(5,20,5):
    L_ss.append([random.randint(min, max) for _ in range(i)])
```

```python
for i in L_ss:
    print(f"Lista bruta:{i}")
    lista_selection = selection_sort(i)
    print(f"Lista ordenada:{lista_selection}\n")
```
