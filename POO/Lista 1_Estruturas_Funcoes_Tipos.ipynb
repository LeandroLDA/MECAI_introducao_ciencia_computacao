## Exercício-exemplo — Filtrando números pares

Crie a função `filtrar_pares(valores)`, que recebe uma lista de números inteiros e retorna uma **nova lista** contendo apenas os valores pares, mantendo a ordem original. A lista recebida não deve ser alterada.

A resposta completa aparece na próxima célula. Execute-a e depois execute os casos de teste para observar o processo de validação.

````python
def filtrar_pares(valores):
    """
    Recebe uma lista de inteiros e retorna uma nova lista apenas com os pares.
    """
    pares = []
    for valor in valores:
        if valor % 2 != 0:
            continue
        pares.append(valor)
    return pares
````

### Entendendo a solução e os testes

A função possui um nome, um parâmetro formal (`valores`), uma *docstring*, um corpo e um valor retornado. A lista `pares` é local à função. O laço percorre a lista recebida, o `continue` ignora os ímpares e `append` inclui cada par na nova lista.

Cada `assert` executa a função com parâmetros reais e compara o resultado com o esperado. Se a condição for falsa, o Python interrompe a célula e mostra um `AssertionError`. Se todos os testes passarem, a mensagem final será exibida.

````python
# Cada assert compara o resultado calculado com o resultado esperado.
entrada = [1, 2, 3, 4, 6]
assert filtrar_pares(entrada) == [2, 4, 6], "A filtragem dos pares está incorreta."
assert entrada == [1, 2, 3, 4, 6], "A função não deve alterar a lista recebida."
assert filtrar_pares([]) == [], "Uma lista vazia deve produzir outra lista vazia."
assert filtrar_pares([-4, -3, 0, 7]) == [-4, 0], "Verifique negativos e zero."

print("✅ Exemplo correto: todos os casos de teste passaram!")
````
