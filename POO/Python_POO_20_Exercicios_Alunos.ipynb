## Exercício-exemplo — Classe `Retangulo`

Implemente uma classe `Retangulo` com:

- atributos de instância `base` e `altura`;
- construtor que rejeita dimensões menores ou iguais a zero com `ValueError`;
- método `area()`, que retorna `base × altura`;
- método `perimetro()`, que retorna `2 × (base + altura)`.

A célula seguinte contém uma solução completa. Execute-a e depois execute os casos de teste.

```python
class Retangulo:
    """Representa um retângulo com dimensões estritamente positivas."""

    def __init__(self, base, altura):
        if base <= 0 or altura <= 0:
            raise ValueError("base e altura devem ser positivas")
        self.base = base
        self.altura = altura

    def area(self):
        return self.base * self.altura

    def perimetro(self):
        return 2 * (self.base + self.altura)
```

### Entendendo o exemplo

`Retangulo` é a classe, enquanto `r1` e `r2` são objetos independentes. O método `__init__` estabelece o estado inicial e o invariante de dimensões positivas. `self` identifica o objeto que recebeu a chamada. Os métodos `area` e `perimetro` formam a interface pública usada pelos testes.

Cada `assert` verifica uma condição esperada. Para o caso inválido, o bloco `try/except` confirma que o construtor sinaliza corretamente o erro com `ValueError`.

```python
# Instanciamos objetos e verificamos estado e comportamento.
r1 = Retangulo(4, 3)
assert r1.base == 4
assert r1.altura == 3
assert r1.area() == 12
assert r1.perimetro() == 14

# Objetos distintos mantêm estados independentes.
r2 = Retangulo(2.5, 2)
assert r2.area() == 5.0
assert r1.area() == 12

# Também verificamos a exceção exigida pelo contrato da classe.
try:
    Retangulo(0, 3)
except ValueError:
    pass
else:
    raise AssertionError("Dimensões inválidas deveriam gerar ValueError.")

print("✅ Exemplo correto: todos os casos de teste passaram!")
```
