## Exercício Prático 1

Implementar o algoritmo de busca sequencial abaixo:

      busca_sequencial(L, chave):
      Para i= 0 até tamanho(L)-1 faça
              Se L[i] == chave, então
                  retornar Verdadeiro  
      retornar Falso

Encontrar valor gerando lista e chave conforme código abaixo:

      min_val = 10
      max_val = 100
      L = []
      for tamanho in range(10, 50, 5):
        L.append([random.randint(min_val, max_val) for _ in range(tamanho)])


      for lista in L:
        print("Lista:", lista)
        chave = random.randint(min_val, max_val)
        print("Chave de busca:", chave)
        resultado = busca_sequencial(lista, chave)
        print("Resultado da busca:", resultado)
        print("-" * 20)


### Solução

```python
def busca_sequencial(L, chave):
    for i in range(len(L)):
        if i == chave:
            return True
    return False
```

```python
# Gerador de dados

min_val = 10
max_val = 100
L = []
for tamanho in range(10, 50, 5):
  L.append([random.randint(min_val, max_val) for _ in range(tamanho)])


for lista in L:
  print("Lista:", lista)
  chave = random.randint(min_val, max_val)
  print("Chave de busca:", chave)
  resultado = busca_sequencial(lista, chave)
  print("Resultado da busca:", resultado)
  print("-" * 20)
```

## Exercício Prático 2
Implementar o algoritmo de busca binária abaixo:

    buscaBinaria(L, inicio, fim, chave)
      Se inicio > fim então
        retorna Falso
      m= (inicio + fim) // 2
      Se L[m] == chave então
        retorna Verdadeiro
      Se L[m] > chave então
        retorna buscaBinaria(L, inicio, meio - 1, chave)
      Senão
        retorna buscaBinaria(L, meio + 1, fim, chave)

Testar para encontrar chaves em listas conforme código abaixo:

      min_val = 10
      max_val = 100
      L = []
      for tamanho in range(10, 50, 5):
        L.append([random.randint(min_val, max_val) for _ in range(tamanho)])


      for lista in L:
        print("Lista:", lista)
        chave = random.randint(min_val, max_val)
        print("Chave de busca:", chave)
        resultado = busca_binária(lista, chave)
        print("Resultado da busca:", resultado)
        print("-" * 20)

### Solução

```python
def buscaBinaria(L, chave):
    L = sorted(L)
    inicio = L[0]
    fim = len(L)-1
    if inicio > fim:
        return False
    m= (inicio + fim) // 2
    if L[m] == chave:
        return True
    elif L[m] > chave:
        return buscaBinaria(L, inicio, m - 1, chave)
    else:
        return buscaBinaria(L, m + 1, fim, chave)
```

```python

# Testar para encontrar chaves em listas conforme código abaixo:

min_val = 10
max_val = 100
L = []
for tamanho in range(10, 50, 5):
  L.append([random.randint(min_val, max_val) for _ in range(tamanho)])


for lista in L:
  print("Lista:", lista)
  chave = random.randint(min_val, max_val)
  print("Chave de busca:", chave)
  resultado = buscaBinaria(lista, chave=chave)
  print("Resultado da busca:", resultado)
  print("-" * 20)
```
