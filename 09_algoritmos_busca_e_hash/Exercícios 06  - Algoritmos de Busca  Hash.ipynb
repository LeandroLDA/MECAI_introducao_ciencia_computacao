Implementar o algoritmo de busca sequencial abaixo:

      busca_sequencial(L, chave):

    buscaBinaria(L, inicio, fim, chave)
      Se inicio > fim então

      m= (inicio + fim) // 2
      Se L[m] == chave então
        retorna Verdadeiro
      Se L[m] > chave então
        retorna buscaBinaria(L, inicio, meio - 1, chave)
      Senão
        retorna buscaBinaria(L, meio + 1, fim, chave)

Testar para encontrar chaves em listas conforme código abaixo:

      max_val = 100
      L = []
      for tamanho in range(10, 50, 5):

      for lista in L:

### Solução

```python
def buscaBinaria(L, inicio=0, fim = None, chave=None):
    if fim == None:
        fim = len(L)-1
    if inicio > fim:
        return False
    m= (inicio + fim) // 2
    if L[m] == chave:
        return True
    elif L[m] > chave:
        return buscaBinaria(L, inicio, m - 1, chave)
    else:
        return buscaBinaria(L, m + 1, fim, chave)
```

## Exercício Prático 3


Implemente uma tabela hash utilizando o método da divisão como função de espalhamento. A tabela hash terá tamanho 10. Os valores inteiros a serem inseridos serão gerados aleatoriamente no intervalo entre 10 e 1000. Ao final, exibir a  quantidade de colisões ocorridas e o conteúdo da tabela hash.

### Solução

```python
class table_hash_linear:
    def __init__(self, tamanho: int=10):
        self.tamanho = tamanho
        self.tabela = [None] * self.tamanho
        self.tot_colisoes = 0

    def func_hash(self, chave: int) -> int:
        return chave % self.tamanho

    def inserir(self, chave: int) -> bool:
        ind = self.func_hash(chave)
        
        if self.tabela[ind] is not None:
            self.tot_colisoes += 1 
        
            tentativas = 0
            while self.tabela[ind] is not None and tentativas < self.tamanho:
                ind = (ind + 1) % self.tamanho
                tentativas += 1

            if tentativas == self.tamanho:
                print("Tabela Cheia.")
                return False
        self.tabela[ind] = chave
        return True
        
        
    def exibir(self) -> None:
        print(f"CONTEUDO DA TABELA (TAMANHO={self.tamanho})\n")
        for i, j in enumerate(self.tabela):
            print(i,":",j)
        print(f"Ocorreram {self.tot_colisoes} colisões.")

    
        
```

```python
if __name__ == "__main__":
    tabela = table_hash_linear(tamanho=10)
    
    insercoes = 8
    valores_p_inserir = [random.randint(10,1000) for i in range(insercoes)]
    print(f"Valores gerados ({insercoes}:{valores_p_inserir})\n")
    
    for valor in valores_p_inserir:
        tabela.inserir(valor)
        
    tabela.exibir()
```

```python
def quicksort(L, low=0, high=None):
    if high is None:
        high = len(L)-1
    if low < high:
        p = particao(L, low, high)
        quicksort(L, low, p-1)
        quicksort(L, p+1, high)
    return L
    
        
def particao(L,low, high):
    i=low
    pivot = L[high]
    for j in range(low, high):
        if L[j] <= pivot:
            L[i], L[j] = L[j], L[i]
            i += 1
    L[i], L[high] = L[high], L[i]
    return i
            
            
lista_ordenada = L
```

## Exercício Prático 6


Plotar gráficos comparando o desempenho do algoritmo de busca binária quando utiliza quick_sort e merge_sort para ordenar em ordem crescente as listas onde a busca será realizada. Considere que cada lista a ser ordenada estpa com elementos dispostos em ordem inversa. Nos dois casos, as listas geradas devem ter tamanho variando de 10 a 100. Utilize 10 repetições para o cálculo do tempo médio.


### Solução

```python
def merge_sort(L):
    if len(L) <= 1:
        return L
    metade = len(L) // 2
    esquerda = merge_sort(L[:metade])
    direita = merge_sort(L[metade:])
    return merge(esquerda, direita)

def merge(esquerda, direita):
    l_aux = []
    i =  j = 0
    while i < len(esquerda) and j < len(direita):
        if esquerda[i] <= direita[j]:
            l_aux.append(esquerda[i])
            i += 1
        else:
            l_aux.append(direita[j])
            j += 1
    l_aux.extend(esquerda[i:])
    l_aux.extend(direita[j:])
    return l_aux
    
```

```python
def quick_e_busca(func):
    def wrapper(lista, chave):
        lista_ordenada = quicksort(lista)
        return func(lista_ordenada, chave)
    return wrapper

def merge_e_busca(func):
    def wrapper(lista, chave):
        lista_ordenada = merge_sort(lista)
        return func(lista_ordenada, chave)
    return  wrapper
```

```python
@quick_e_busca
def busca_quick(lista, chave):
    return busca_binaria(lista, chave)

@merge_e_busca
def busca_merge(lista, chave):
    return busca_binaria(lista, chave)
```

```python
def plotar_comparacao(dados_grafico, titulo="Comparação de Desempenho"):
    plt.figure(figsize=(8, 5))
    for config in dados_grafico:
        plt.scatter(config['eixo_x'], config['tempos'], color=config['cor'], label=config['label'])
    
    plt.xlabel("Tamanho da Lista")
    plt.ylabel("Tempo Médio (s)")
    plt.title(titulo)
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.show()
```

## Exercício Prático 9


Implemente uma tabela hash utilizando o método de folding como função de espalhamento. A tabela hash terá tamanho 100. Para tratar colisões, utilize sondagem quadrática como método de endereçamento aberto. Ao final, exibir a  quantidade de colisões ocorridas e o conteúdo da tabela hash.

Utilize as mesmas chaves geradas na Exercício Prático 7.

### Solução

```python
class table_hash_quad:
    def __init__(self, tamanho: int=10):
        self.tamanho = tamanho
        self.tabela = [None] * self.tamanho
        self.tot_colisoes = 0

    
    def func_hash(self, chave: int) -> int:
        str_chave = str(chave)
        recorte = 2
        soma = 0
        
        for i in range(0, len(str_chave), recorte):
            pedaco = str_chave[i:i+recorte]
            soma += int(pedaco)
        return soma % self.tamanho
    

    def inserir(self, chave: int) -> bool:
        indice = self.func_hash(chave)
        
        if self.tabela[indice] is None:
            self.tabela[indice] = chave
            return True

        self.tot_colisoes += 1 
        tentativas = 1
        while tentativas < self.tamanho:
            ind = (indice + tentativas**2) % self.tamanho
     
            if self.tabela[ind] is None:
                self.tabela[ind]  = chave
                return True
            tentativas += 1
            
        print("Tabela Cheia")
        return False
        
        
    def exibir(self) -> None:
        print(f"CONTEUDO DA TABELA (TAMANHO={self.tamanho})\n")
        for i, j in enumerate(self.tabela):
            print(i,":",j)
        print(f"Ocorreram {self.tot_colisoes} colisões.")
```

```python
tabela_ex9 = table_hash_quad(tamanho=100)
for chave in random_keys:
    tabela_ex9.inserir(chave)
    
    
tabela_ex9.exibir()
```
